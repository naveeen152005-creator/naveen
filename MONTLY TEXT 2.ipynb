```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(df['Quantity'])

plt.title('Outlier Detection - Quantity')
plt.xlabel('Quantity')
plt.show()
```

```python
#8 create at least 4 appropriate visualizations 
#barchart 

category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
#histogram
import matplotlib.pyplot as plt

plt.hist(df["Sales"], bins=10)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
#boxplot
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(y=df["Sales"])

plt.title("Sales Boxplot")
plt.ylabel("Sales")
plt.show()
```

```python
#piechart
category_sales = df.groupby("Category")["Sales"].sum()

plt.pie(category_sales, labels=category_sales.index, autopct="%1.1f%%")

plt.title("Sales by Category")
plt.show()
```

```python
#linechart
plt.figure(figsize=(8, 4))
sns.lineplot(
    data=df,
    x='Quantity',
    y='Unit_Price',
    marker='o'
)
plt.title('quantity vs unit price')
plt.xlabel('Quantity')
plt.ylabel('Unit_Price')
plt.show()
```

```python
#PART C - BUSSINESS REQUIREMENTS

#1) 

sales_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(sales_category)
print("Highest Sales Category:", sales_category.idxmax())
```

```python
#2)
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(region_sales)
print("Best Region:", region_sales.idxmax())
print("Worst Region:", region_sales.idxmin())
```
