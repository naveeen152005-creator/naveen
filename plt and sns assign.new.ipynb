```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_excel(r"C:\Users\computer\Downloads\Sales_Visualization_Practice_34399618.xlsx")

print(df.head())
print(df.columns.tolist())
```

```python
sales = df.groupby("Category")["Sales"].sum()
sales.plot(kind="bar")
plt.title("Sales by Category")
plt.show()
```

```python
sales = df.groupby("Region")["Sales"].sum()
sales.plot(kind="bar")
plt.title("Sales by Region")
plt.show()
```

```python
sales = df.groupby("Customer_Segment")["Sales"].sum()
sales.plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.show()
```

```python
sales = df.groupby("Month")["Sales"].sum()
sales.plot(kind="line", marker="o")
plt.title("Monthly Sales Trend")
plt.show()
```

```python
profit = df.groupby("Month")["Profit"].sum()
profit.plot(kind="line", marker="o")
plt.title("Monthly Profit Trend")
plt.show()
```

```python
sales = df.groupby("Product")["Sales"].sum()
top5 = sales.nlargest(5).sort_values()
top5.plot(kind="barh")
plt.title("Top 5 Products by Sales")
plt.show()
```

```python
sales = df.groupby("City")["Sales"].sum()
sales.plot(kind="bar")
plt.title("Sales by City")
plt.show()
```

```python
plt.scatter(df["Quantity"], df["Sales"])
plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.show()
```

```python
table = df.pivot_table(
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum"
)
sns.heatmap(table, annot=True, fmt=".0f")
plt.show()
```

```python
plt.hist(df["Sales"].dropna(), bins=15)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```
