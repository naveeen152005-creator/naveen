```python
#1. Sorting – Ascending
numbers = [45, 12, 78, 3, 29, 56, 8]
result = sorted(numbers, key=lambda x: x)
print(result)
```

```python
#2. Sorting – By String Length
words = ["python", "ai", "machine", "data", "learning", "ml"]
result = sorted(words, key=lambda x: len(x))
print(result)
```

```python
#3. Sorting – Descending
numbers = [15, 42, 7, 89, 23, 64, 31]
result = sorted(numbers, key=lambda x: x, reverse=True)
print(result)

```

```python
4.#Filtering – Even Numbers
numbers = [12, 7, 25, 40, 33, 18, 9, 50]
result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)
```

```python
#5.Filtering – Numbers Greater Than 50
numbers = [25, 67, 12, 89, 45, 73, 34, 91]
result = list(filter(lambda x: x > 50, numbers))
print(result)
```

```python
#6. Mapping – Square Numbers
numbers = [2, 4, 6, 8, 10]
result = list(map(lambda x: x * x, numbers))
print(result)
```

```python
#7. Mapping – Convert to Positive
numbers = [-10, 5, -7, 12, -3, 8]
positive = list(map(lambda x: abs(x), numbers))
print(positive)
```

```python
#8. Reducing – Sum
from functools import reduce
numbers = [10, 20, 30, 40, 50]
total = reduce(lambda a, b: a + b, numbers)
print(total)
```

```python
#9. Reducing – Maximum Number
from functools import reduce
numbers = [45, 12, 89, 34, 67, 23]
maximum = reduce(lambda a, b: a if a > b else b, numbers)
print(maximum)
```
