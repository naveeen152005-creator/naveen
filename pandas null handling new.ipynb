```python
import pandas as pd
import numpy as np

```

```python

# Create data
data = {
    "Name": ["Anu", "Rahul", "Akhil", "Meera", "Neha", "Arjun"],
    "Age": [23, np.nan, 25, 24, np.nan, 27],
    "City": ["Kochi", "Calicut", np.nan, "Kochi", "Thrissur", "Kochi"],
    "Salary": [25000, 30000, 35000, np.nan, 28000, 40000],
    "Department": ["IT", "HR", "IT", "Finance", "HR", np.nan]
}

```

```python
# Create DataFrame
df = pd.DataFrame(data)

```

```python
# Display DataFrame
print("Original DataFrame:")
print(df)

```

```python
# Check number of null values in each column
print("\nNull values in each column:")
print(df.isnull().sum())

```

```python
# Display rows containing null values
print("\nRows containing null values:")
print(df[df.isnull().any(axis=1)])

```

```python
# Replace null values in Age with mean age
df["Age"] = df["Age"].fillna(df["Age"].mean())

```

```python
# Replace null values in City with "Unknown"
df["City"] = df["City"].fillna("Unknown")

```

```python
# Replace null values in Salary with median salary
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

```

```python
# Replace null values in Department with "Not Assigned"
df["Department"] = df["Department"].fillna("Not Assigned")

```

```python
# Check null values again
print("\nNull values after cleaning:")
print(df.isnull().sum())

```

```python
# Display final cleaned DataFrame
print("\nFinal Cleaned DataFrame:")
print(df)
```
