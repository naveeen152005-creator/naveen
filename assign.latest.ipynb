```python
#1. Class and Object
class Student:
    name = "Rahul"
    age = 22
s = Student()
print("Name:", s.name)
print("Age:", s.age)
```

```python
#2. Class Method
class Car:
    brand = "Toyota"
    model = "Innova"
    def display(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
c = Car()
c.display()
```

```python
#3. Constructor (__init__)
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.salary = salary
e = Employee("Anu", "IT", 30000)
print("Name:", e.name)
print("Department:", e.department)
print("Salary:", e.salary)
```

```python
#4 multiple Objects
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
a1 = BankAccount("Arun", 25000)
a2 = BankAccount("Rahul", 40000)
a3 = BankAccount("Neha", 35000)
print("Account Holder:", a1.account_holder)
print("Balance:", a1.balance)
print("Account Holder:", a2.account_holder)
print("Balance:", a2.balance)
print("Account Holder:", a3.account_holder)
print("Balance:", a3.balance)
```

```python
#5.Method with Calculation
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
r = Rectangle(10, 5)
print("Area:", r.area())
```

```python
#Exception Handling

#6. 6. Division by Zero
try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
#7. Invalid Input
try:
    num = int(input("Enter a number: "))
    print("Number:", num)
except ValueError:
    print("Invalid input! Please enter a number.")
```

```python
#8. Multiple Exceptions
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print(a / b)
except ValueError:
    print("Please enter numbers only.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
#9. List Index Exception
numbers = [10, 20, 30, 40, 50]
try:
    index = int(input("Enter index: "))
    print(numbers[index])
except IndexError:
    print("Invalid index")
```

```python
#10. Raise Your Own Exception
try:
    age = int(input("Enter age: "))
    if age < 18:
        raise Exception("You must be 18 or older.")
    print("You are eligible.")
except Exception as e:
    print("Exception:", e)
```
