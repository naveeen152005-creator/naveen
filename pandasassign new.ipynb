```python
import pandas as pd

data = {
    "Name": [
        " Rahul ", "PRIYA", "arjun", " Sneha ", "rahul",
        "Akhil", None, "Meera", "PRIYA"
    ],

    "Age": [
        "22", "23", None, "21", "22",
        "24", "25", "23", None
    ],

    "Course": [
        "Data Science", "data science", "DATA SCIENCE",
        "Python", None, "Python", "Data Science",
        "python", "data science"
    ],

    "Salary": [
        "25000", "30000", "28000", None, "25000",
        "32000", "35000", "30000", "30000"
    ],

    "Join_Date": [
        "2025-01-10", "15/01/2025", "2025-02-05",
        None, "2025-01-10", "20/02/2025",
        "2025-03-01", "invalid", "15/01/2025"
    ]
}
```

```python
df = pd.DataFrame(data)

```

```python
print(df)
```

```python
#string cleaning 
# Clean Name
df["Name"] = df["Name"].str.strip()
df["Name"] = df["Name"].str.title()

# Clean Course
df["Course"] = df["Course"].str.strip()
df["Course"] = df["Course"].str.title()

print(df)
```

```python
print(df.isnull().sum())
```

```python
#fill missing values
# Clean Name
df["Name"] = df["Name"].str.strip()
df["Name"] = df["Name"].str.title()

# Clean Course
df["Course"] = df["Course"].str.strip()
df["Course"] = df["Course"].str.title()

print(df)
```

```python
#duplicate removal 
# Find duplicate rows
print("Duplicate rows:")
print(df[df.duplicated()])

# Count duplicate rows
print("Number of duplicates:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print(df)
```

```python
df["Salary"] = df["Salary"].astype(float)
```

```python
df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    errors="coerce",
    dayfirst=True
)

print(df)
```
