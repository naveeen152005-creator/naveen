```python
#. Create an Array
import numpy as np
arr = np.arange(1, 11)
print(arr)
```

```python
#2. Array Operations
import numpy as np
a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
#3. Maximum, Minimum & Average
import numpy as np
marks = np.array([85, 72, 91, 65, 78, 88])
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Average:", np.mean(marks))
```

```python
#4. Even Numbers
import numpy as np
arr = np.arange(1, 21)
even = arr[arr % 2 == 0]
print(even)
```

```python
#5. Reshape an Array
import numpy as np
arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)
print(matrix)
```
